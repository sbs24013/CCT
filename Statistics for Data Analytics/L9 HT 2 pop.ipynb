# A different example: 50 startups

```python
import pandas as pd
import numpy as np

dataset = pd.read_csv("./datasets/50_Startups_dataset.csv")
dataset.head()
```

```python
X1 = dataset['Administration']
X2 = dataset['Marketing Spend']
```

```python
t_test_startups = stats.ttest_ind(X1, X2, equal_var = False)
t_test_startups
```

```python
# As pvalue is lower than alpha, we reject the null hypothesis
# **Interpretation**: With a 5% significant level, there is enough evidence to say that there are differences between the expenses of Administration and Marketing
```
