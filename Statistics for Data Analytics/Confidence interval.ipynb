```python
import pandas as pd
import numpy as np
import scipy.stats as st
```

# CI for

```python
# 문제에서 제공하는 변수

sample_mean = 2_000 # x bar
n = 500 # sample size
std_population = 300 # sigma
alpha = 0.05 # significance lavel or (1 - confidence level)
```

```python
# 신뢰구간 계산

confidence_interval = st.norm.interval(confidence= 1 - alpha, loc = sample_mean, scale = std_population/np.sqrt(n))
confidence_interval
```

# CI for population mean when sigma is unknown

```python
x_bar = 250
std_sample = 25
n1 = 300
alpha3 = 0.03

# Creating the confidence interval at 97% confidence

t_condifidence_interval = st.t.interval(confidence = 1 - alpha3, df = n1-1, loc=x_bar, scale=std_sample / np.sqrt(n1-1))

t_condifidence_interval
```
