## Web Query

https://api.open-meteo.com/v1/forecast?latitude=52.52&longitude=13.41&current=temperature_2m&hourly=temperature_2m,rain,wind_speed_10m,temperature_80m 

- protocol: https://
- domain: api.open-meteo.com
- endpoint: v1/forecase
- ? : start of query
- key=value: latitude=52.52
- parameters are divided by &

```python
import requests
import json
```

```python
url = "https://api.open-meteo.com/v1/forecast?latitude=52.52&longitude=13.41&current=temperature_2m&hourly=temperature_2m,rain,wind_speed_10m,temperature_80m"
```

```python
res = requests.get(url)
```
