```python
def csv_to_sql(path="", table_name="", schema = "", conn = conn, if_exists='append'):
  df = pd.read_csv(path)
  result = df.to_sql(name=table_name, con=conn, index=False, schema=schema, if_exists=if_exists)
  print(result)

csv_to_sql(path = "./datasets/Pepsico.csv", schema="companies", table_name = "all_companies")
```
