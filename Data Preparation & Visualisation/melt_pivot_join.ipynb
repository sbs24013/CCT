```python
temps = pd.DataFrame({
  'city': ["New_York","Athens","Chicago","Miami"],
  "day1": [25,36,18,41],
  "day2": [26,36,16,48],
  "day3": [28,38,18,48],
  "day4": [25,35,15,45]
})

# """
#   "New_York": [25,26,28,25],
#   "Athens": [36,36,38,35],
#   "Chicago": [18,16,18,15],
#   "Miami": [41,46,48,45]
# """

temps
```

```python
temps_beta = temps.melt(id_vars=["city"], var_name="date", value_name="temperature")
```

```python
temps_beta
```

```python
temps_gamma = temps_beta.pivot(index="date", columns="city")
temps_gamma
```

```python
temps_gamma = temps_beta.pivot(index="city", columns="date", values="temperature")
temps_gamma
```

```python
temps_gamma = temps_beta.pivot(index="city", columns="date", values="temperature").reset_index()
temps_gamma
```
